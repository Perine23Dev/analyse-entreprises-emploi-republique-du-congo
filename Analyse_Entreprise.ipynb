# Partie B : Nettoyage des données

Après l'exploration de la base, plusieurs anomalies, incohérences et valeurs manquantes ont été identifiées. Le nettoyage consiste à corriger ou traiter ces problèmes afin de disposer de données cohérentes et fiables pour l'analyse des entreprises et de leur contribution à l'emploi des Congolais.

Une attention particulière est accordée aux variables directement liées à la mesure de l'emploi, à l'identification des entreprises, à leur localisation, à leur activité et à leur taille.

## Opérations de nettoyage des données

- Changement des types de données des colonnes non correspondants
- Suppression des espaces inutiles avant et après les valeurs
- Traitement des valeurs manquantes
- Traitement des valeurs aberrantes et incohérentes
- Harmonisation de la casse des valeurs textuelles
- Correction des différentes variantes d'orthographe
- Complétion des informations manquantes lorsque cela est possible
- Harmonisation des modalités de la colonne `Genre`
- Harmonisation des modalités des variables catégorielles
- Vérification et correction de la cohérence entre les colonnes liées à l'emploi
- Vérification de la cohérence entre le nombre total de salariés et le nombre d'employés congolais
- Vérification et correction des valeurs numériques incohérentes
- Vérification des années de création inhabituelles
- Harmonisation des modalités de la colonne `Département`
- Harmonisation des modalités liées aux activités, branches et secteurs
- Vérification de la cohérence des catégories d'entreprises


````python
conversion = ["Age",
"capital",
"chiffre_aff_mo_annuel",
"chiffre_aff_mo_annuel_1",
"chiffre_aff_mo_annuel_2",
"nbre_empl_congolais",
"salarie",
"part_capital_etranger" ]
for col in conversion:
  df[col] = df[col].astype(str).str.strip().str.replace(r'\s+', '', regex=True)
  df[col] = pd.to_numeric(df[col], errors='coerce').round().astype("Int64")
````

````python
# Nettoyage en capitalize des colonnes indiquées dans la liste
space_clean = ["Département", "Activité déclarée", "Activité au sens de la nomenclature OHADA", "Statut juridique", "Genre", "Nationalité", "Branche", "Secteur"]
for col in space_clean:
    df[col] = df[col].astype(str).str.capitalize()
````

```markdown
## Analyse du profil des entreprises qui génèrent le plus d'emplois

Dans cette section, nous créons un groupe d'entreprises à fort impact sur l'emploi (les **Grands Employeurs**, définis par exemple comme le Top 5% des entreprises ayant le plus grand nombre de salariés)
et nous analysons leurs caractéristiques par rapport au reste des entreprises.
```

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency, ttest_ind

# Assurer que la colonne salarie est bien numérique
df['salarie'] = pd.to_numeric(df['salarie'], errors='coerce')

# Détermination du seuil pour le Top 5% des employeurs
seuil_top_employeurs = df['salarie'].quantile(0.95)
print(f"Seuil pour faire partie du Top 5% des employeurs : {seuil_top_employeurs:.1f} salariés")

# Étiquetage des entreprises (Grands Employeurs vs Autres)
df['profil_employeur'] = np.where(df['salarie'] >= seuil_top_employeurs, 'Grand Employeur', 'Petit/Moyen Employeur')

# Distribution globale
print("\nRépartition des entreprises par profil d'employeur :")
print(df['profil_employeur'].value_counts(dropna=False))
````

````python
# 1. Croisement avec le Secteur d'activité
print("=== Profil par Secteur d'Activité ===")
secteur_profil = pd.crosstab(df['Secteur'], df['profil_employeur'], normalize='columns') * 100
display(secteur_profil.round(2))

# Test statistique du Chi-Deux pour le Secteur
chi2, p, dof, ex = chi2_contingency(pd.crosstab(df['Secteur'], df['profil_employeur']))
print(f"Test du Chi-deux (Secteur vs Profil) : p-value = {p:.4e}")

# 2. Croisement avec le Statut Juridique
print("\n=== Profil par Statut Juridique ===")
statut_profil = pd.crosstab(df['Statut juridique'], df['profil_employeur'], normalize='columns') * 100
display(statut_profil.round(2))

# Test statistique du Chi-Deux pour le Statut juridique
chi2_stat, p_stat, _, _ = chi2_contingency(pd.crosstab(df['Statut juridique'], df['profil_employeur']))
print(f"Test du Chi-deux (Statut vs Profil) : p-value = {p_stat:.4e}")
````

````python
# 3. Analyse de l'âge moyen des entreprises selon leur profil
print("=== Analyse de l'âge des entreprises ===")
age_stats = df.groupby('profil_employeur')['age'].agg(['mean', 'median', 'std', 'count']).reset_index()
display(age_stats.round(2))

# Test de Student (T-test) pour comparer l'âge moyen
groupe_grands = df[df['profil_employeur'] == 'Grand Employeur']['age'].dropna()
groupe_petits = df[df['profil_employeur'] == 'Petit/Moyen Employeur']['age'].dropna()
t_stat, p_val_t = ttest_ind(groupe_grands, groupe_petits, equal_var=False)
print(f"Test T de Student (Comparaison des âges moyens) : p-value = {p_val_t:.4e}")
````

````python
import prince
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Sélection des variables catégorielles clés pour l'ACM
features_mca = ['Secteur', 'Statut juridique', 'profil_employeur']

# Filtrage des lignes sans valeurs manquantes sur ces variables pour l'ACM
df_mca = df[features_mca].dropna().copy()

# Initialisation et ajustement de l'ACM
mca = prince.MCA(
    n_components=2,
    n_iter=10,
    copy=True,
    check_input=True,
    engine='sklearn',
    random_state=42
)
mca = mca.fit(df_mca)

# Récupération des coordonnées des modalités
coordonnees_modalites = mca.column_coordinates(df_mca)
coordonnees_modalites.columns = ['Axe 1', 'Axe 2']

# Récupération sécurisée et conversion forcée en numérique du pourcentage d'inertie pour les axes
try:
    # Version prince moderne
    val1 = mca.percentage_of_variance_.iloc[0] if isinstance(mca.percentage_of_variance_, pd.Series) else mca.percentage_of_variance_[0]
    val2 = mca.percentage_of_variance_.iloc[1] if isinstance(mca.percentage_of_variance_, pd.Series) else mca.percentage_of_variance_[1]
    inertie_axe1 = float(val1) * 100
    inertie_axe2 = float(val2) * 100
except Exception:
    try:
        val1 = mca.eigenvalues_summary.iloc[0, 1]
        val2 = mca.eigenvalues_summary.iloc[1, 1]
        inertie_axe1 = float(val1) * 100
        inertie_axe2 = float(val2) * 100
    except Exception:
        inertie_axe1 = 0.0
        inertie_axe2 = 0.0

# Visualisation de la carte des modalités
plt.figure(figsize=(12, 10))
sns.scatterplot(
    x='Axe 1',
    y='Axe 2',
    data=coordonnees_modalites,
    s=150,
    color='darkblue',
    alpha=0.8
)

# Ajout des étiquettes des modalités avec un décalage pour la lisibilité
for i, txt in enumerate(coordonnees_modalites.index):
    plt.annotate(
        txt,
        (coordonnees_modalites['Axe 1'].iloc[i], coordonnees_modalites['Axe 2'].iloc[i]),
        textcoords="offset points",
        xytext=(0,10),
        ha='center',
        fontsize=10,
        weight='bold'
    )

plt.axhline(0, color='gray', linestyle='--', linewidth=0.8)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.8)
plt.title("Analyse des Correspondances Multiples (ACM) - Carte des Profils d'Entreprises", fontsize=14, pad=15)
plt.xlabel(f"Axe 1 ({inertie_axe1:.2f}% d'inertie)", fontsize=11)
plt.ylabel(f"Axe 2 ({inertie_axe2:.2f}% d'inertie)", fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
````
