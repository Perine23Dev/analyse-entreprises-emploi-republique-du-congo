```python
import pandas as pd

chemin = '/content/drive/MyDrive/Groupe_3 Akieni/Dataset/Entreprise_congo.xlsx'

df = pd.read_excel(chemin)
```

```python
cols = df.columns[-5:]

cols_to_convert_map = {
    cols[0]: 'Int64', # 'capital'
    cols[1]: 'Int64', # 'part_capital_etranger'
    cols[2]: 'float64', # 'salarie'
    cols[3]: 'float64', # 'nbre_empl_congolais'
    cols[4]: 'float64' # 'chiffre_aff_mo_annuel'
}

# Identify columns that are 'object' type and need cleaning before conversion
# Based on df.info(), 'capital' and 'chiffre_aff_mo_annuel' are likely candidates.
object_cols_in_map = [col for col, dtype in cols_to_convert_map.items() if df[col].dtype == 'object']

for col_name in object_cols_in_map:
    # Ensure the column is treated as string before applying string methods
    # Then remove spaces and replace commas with dots
    df[col_name] = df[col_name].astype(str).str.replace(' ', '', regex=False)
    df[col_name] = df[col_name].str.replace(',', '.', regex=False)
    # Convert to numeric, coercing errors to NaN
    df[col_name] = pd.to_numeric(df[col_name], errors='coerce')

# Now apply the final astype conversion, which will work with cleaned and numeric-like data
df = df.astype(cols_to_convert_map)
```
